### For Stationarity

```python
from statsmodels.tsa.stattools import adfuller

# Perform the Augmented Dickey-Fuller test
adf_test = adfuller(df2['Close'])

print('ADF Test Results:')
print(f'ADF Statistic: {round(adf_test[0],3)}')
print(f'p-value: {round(adf_test[1],3)}')
print('Critical Values:')

for key, value in adf_test[4].items():
    print(f'{key}: {round(value,3)}')
```

```python
adf_test[4]
```

```python
from statsmodels.tsa.stattools import kpss

# Perform the KPSS test
kpss_test = kpss(df2['Close'], regression='ct') # 'c' for constant (level stationarity), 'ct' for trend

print( 'KPSS Test Results:')
print(f'KPSS Statistic: {round(kpss_test[0],2)}')
print(f'p-value: {kpss_test[1]}')
print( 'Critical Values:')

for key, value in kpss_test [3].items():
    print(f' {key} : {value}')
```

### For checking stationarity

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ks_2samp

# Generating synthetic data for demonstration
np.random.seed(0)
n = 500

# Strictly stationary series: normally distributed random noise
strict_stationary_series = np. random.normal(0, 1, n)

# Non-strictly stationary series: changing variance
non_strict_stationary_series = np.concatenate([
    np.random.normal(0, 1, n // 2), 
    np.random.normal(0, 2, n // 2)
])

# Function to split the series into two halves and perform the K-S test
def ks_test_stationarity(series):
    split = len(series) // 2
    series_first_half = series[:split]
    series_second_half = series[split:]
    stat, p_value = ks_2samp(series_first_half, series_second_half)
    return stat, p_value

# Perform the K-S test on strictly stationary series
ks_stat_strict, ks_pvalue_strict = ks_test_stationarity(strict_stationary_series)

# Perform the K-S test on non-strictly stationary series
ks_stat_non_strict, ks_pvalue_non_strict = ks_test_stationarity(non_strict_stationary_series)
```

### Plotting the Stationarity

```python
# Print the results
print("K-S Test on Strictly Stationary Series:")
print(f"Statistic: {ks_stat_strict:.4f}, p-value: {ks_pvalue_strict:.4f}")

print("\nK-S Test on Non-Strictly Stationary Series:")
print(f"Statistic: {ks_stat_non_strict:.4f}, p-value: {ks_pvalue_non_strict:.4f}")

# Plotting the series
plt.figure(figsize=(14, 6))

plt.subplot(2, 1, 1)
plt.plot(strict_stationary_series, color = 'g')
plt.title('Strictly Stationary Series')

plt.subplot(2, 1, 2)
plt.plot(non_strict_stationary_series, color = 'r')
plt.title('Non-Strictly Stationary Series')

plt.tight_layout()
plt.show()
```

### Defining the function for ADF and KPSS

```python
def adf(df):
    adf = adfuller(df)
    print("ADF Results: ")
    print(f'ADF statistics: {round(adf[0],3)}')
    print(f'P-value: {round(adf[1],3)}')
    print(f'Critical Values: {adf[4]}')

def kpss_res(df):
    kpss_test = kpss(df, regression = 'ct')
    print("KPSS Results" )
    print(f'KPSS Stats: {round(kpss_test[0],3)}')
    print(f'P value: {round(kpss_test[1],3)}')
    print(f'Critical Value: {kpss_test[3]}')
```

```python
adf(df2['Close'])
```

```python
kpss_res(df2['Close'])
```

### Summary:
1. In ADF, if p < 0.05 that means the data is Stationary.
2. In KPSS, if p > 0.05 that means the data is Stationary.

### With First order differencing, Data is stationary

```python
stationary_data1 = df2['Close'].diff().dropna()
stationary_data1
```

```python
stationary_data = df2.Close.diff().dropna()
```

```python
stationary_data
```

### Splitting the training data and testing data

```python
train_data, test_data = stationary_data[:-30], stationary_data[-30:]
```

```python
train_data
```

```python
test_data
```

### AR Model

```python
from statsmodels.tsa.ar_model import AutoReg
```

```python
model = AutoReg(train_data, lags = 30)
model_fit = model.fit()
```

```python
y_pred = model_fit.predict(start = len(train_data), end = len(train_data) + len(test_data)-1, dynamic = False)
```

```python
y_pred
```

```python
test_data
```

```python
plt.figure(figsize = (10,5))
plt.plot(test_data.index, test_data, label = 'test_data')
plt.plot(test_data.index, y_pred, color = 'r', label = 'Prediction')
```

```python
from sklearn.metrics import mean_squared_error
RMSE = np.sqrt(mean_squared_error(test_data, y_pred))
```

```python
RMSE
```

### MA Model

```python
from statsmodels.tsa.arima.model import ARIMA
arima_model = ARIMA(train_data, order = (0,0,30)) # order = (p,d,q), p is autoregressive, differences, and moving averge
arima_model_fit = arima_model.fit()
```

```python
arima_pred = arima_model_fit.predict(start = len(train_data), end = len(train_data) + len(test_data)-1, dynamic = False)
```

```python
plt.figure(figsize = (10,5))
plt.plot(test_data.index, test_data, label = 'test_data')
plt.plot(test_data.index, arima_pred, color = 'r', label = 'Prediction')
```

```python
from sklearn.metrics import mean_squared_error
RMSE = np.sqrt(mean_squared_error(test_data, arima_pred))
```
